```python
# Atividade 1 - Referências
a = [10, 20, 30]
b = a

b.append(40)

mesmo_objeto = id(a) == id(b)

#cópia para demonstrar a diferença
c = a.copy()
a.append(50)

print("---> 1. Identidade de objetos (id) <---")
print("Valor de a:", a, "id(a):", id(a))
print("Valor de b:", b, "id(b):", id(b))
print("a e b representam o mesmo objeto?", mesmo_objeto)
#Ambas as variáveis guardam a referência para o mesmo endereço de memória

print("\n---> 2. Por quê? <---")
x = 60
id_x_antes = id(x)
x = x + 10
print("id da variável 'x' original:", id_x_antes, "--> novo id da variável 'x' após modificação:", id(x))
print("Valores inteiros são imutáveis, então a modificação do objeto original gera um novo objeto.")

print("\n---> 3. Referência vs Cópia de lista <---")
print("Original (a):   ", a, "id(a):", id(a))
print("Referência (b): ", b, "id(b):", id(b), "Sofreu alteração juntamente com 'a'")
print("Cópia (c):      ", c, "id(c):", id(c), "Objeto novo, alocado em outro endereço")
# Um novo objeto do tipo lista seria criado na memória, contendo uma cópia dos elementos.
# Como os endereços (id) passam a ser diferentes, modificar a lista 'b' não afeta a lista 'a'
```

```python
# Atividade 2 - Construindo uma cadeia

class Node:
  def __init__(self, valor):
    self.valor = valor
    self.proximo = None

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
  print(atual.valor)
  atual = atual.proximo
```
