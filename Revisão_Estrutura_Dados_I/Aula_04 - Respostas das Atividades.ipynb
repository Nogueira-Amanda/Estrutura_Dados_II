##Estruturas LIFO e FIFO
`Fundamentos e implementação de Pilhas e Filas`

*Teoria e Metáfora:* Qual a melhor lógica para compreender.
* Pilha (LIFO): O último elemento a ser inserido é o último a ser removido.
Uma estrutura de dados restrita onde a inserção e a remoção de elementos ocorre estritamente por uma única extremidade (chamada TOPO).
Em python, utilizamos a estrutura cçassica de listas para simular pilhas com alta eficiência de ponta.
Método `append()` adiciona ao final (atua como o topo da pilha) e `pop()` sem argumentos remove e retorna esse extato último elemento.

* Fila (FIFO): o primeiro inserido é o primeiro a ser removido.
Utilizamos `collections.deque`.\
Listas comuns em python são ineficientes para filas, pois remover do início exige o deslocamento de todos os outros elementos na memória.

**Exercício 2: O sistema de impressão:**
Objetivo: implementar uma fila para gerenciar um spooler de impressora.
Requisitos: Cada novo documento é enfileirado. O sistema deve garantir que o documento impresso seja sempre o mais antigo aguardando na fila.



```python
class EditorTexto:
    def __init__(self):
        self.historico = []  # <-- Essa é a pilha para armazenar o histórico de texto digitado

    def executar(self, acao):
        self.historico.append(acao)  # <-- Insere no topo da pilha
        print(f"-> Ação '{acao}' adicionada ao histórico.")

    def desfazer(self):
        if not self.historico: # <-- Verifica se a pilha está vazia
            print("-> Nada para desfazer.")
            return None

        ultimaAcao = self.historico.pop() # <-- Para retirar o elemento da pilha
        print("Ação desfeita:", ultimaAcao)
        return ultimaAcao

    def exibir(self): # <-- Para exibir a pilha para o usuário
        print(f"\nAqui está o hitórico da Pilha: {self.historico}\n")


# ---> Menu interativo do usuário <---

editor = EditorTexto()

while True:
    print("-- MENU --")
    print("1 - Executar nova ação")
    print("2 - Desfazer última ação")
    print("3 - Sair")

    opcao = input("Escolha a opção desejada: ")

    if opcao == "1":
        texto = input("Digite a ação desejada: ")
        if texto:
            editor.executar(texto)
            editor.exibir()
        else:
            print("[!!] A entrada não pode ser vazia [!!]")

    elif opcao == "2":
        editor.desfazer()
        editor.exibir()

    elif opcao == "3":
        print("Programa encerrado com sucesso...")
        break

    else:
        print("Opção inválida. Tente novamente.\n")
```

```python
from collections import deque  # <-- Utilizando a coleção para filas

class SpoolerImpressora:
    def __init__(self):
        self.fila_impressao = deque()  # <-- Fila para guardar os documentos para impressão

    def adicionar_documento(self, documento):
        self.fila_impressao.append(documento)  # <-- Os novos elementos são implemntados no final da fila
        print(f"-> Documento '{documento}' adicionado à fila.")

    def imprimir_proximo(self):
        if not self.fila_impressao:
            print("-> Nenhum documento na fila para imprimir.")
            return None

# Remove e retorna o primeiro elemento inserido conforme lógica FIFO
        proximo = self.fila_impressao.popleft()
        print(f"-> Imprimindo: '{proximo}'...")
        return proximo

    def exibir(self): # <-- Exibe a fila conforme ordem de inserção
        documentos = list(self.fila_impressao)
        print("\nDocumentos aguardando na fila:\n", documentos)


# --- Interação com o usuário ---
spooler = SpoolerImpressora()

while True:
    print("--- SPOOLER DE IMPRESSÃO ---")
    print("1 - Adicionar novo documento")
    print("2 - Imprimir o próximo documento")
    print("3 - Sair")

    opcao = input("Escolha a opção desejada: ")

    if opcao == "1":
        doc = input("Digite o nome do documento: ")
        if doc:
            spooler.adicionar_documento(doc)
            spooler.exibir()
        else:
            print("[!!] Nome do documento não pode ser vazio [!!]")

    elif opcao == "2":
        spooler.imprimir_proximo()
        spooler.exibir()

    elif opcao == "3":
        print("Programa encerrado com sucesso.")
        break

    else:
        print("Opção inválida. Tente novamente.\n")
```
